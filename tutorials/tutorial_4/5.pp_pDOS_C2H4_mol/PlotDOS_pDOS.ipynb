# Plotting DOS and pDOS for C$_{2}$H$_{4}$

This simple Jupyter notebook allows you to plot the DOS and pDOS produced by CONQUEST and the post-processing.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## DOS

First we plot the total DOS

```python
# Specify & as a comment as the file ends with it
# column 2 shows integrated DOS if you want it
en, dos = np.loadtxt("DOS.dat",unpack=True,comments=['#','&'],usecols=(0,1))
```

```python
plt.plot(en,dos)
plt.xlim(-10,10)
plt.ylim(0,30)
plt.xlabel("Energy (eV)")
plt.ylabel("DOS (arbitrary units)")
#plt.savefig("Total_DOS.png",dpi=300)
plt.show()
```

## Projected DOS

Using the files ``Atom0000001DOS_l.dat`` we can plot the projected DOS.  Note that we specified that the pDOS should be resolved by $l$ giving five columns in the file (energy, total pDOS, l=0, l=1, l=2).  The parameter used was ``Process.pDOS_l_resolved T``; the pDOS can be further resolved by $m$ by setting ``Process.pDOS_lm_resolved T`` if desired.

```python
# Specify & as a comment as the file ends with it
# columns are energy, total pDOS, l=0, l=1, l=2
en, pdos, pdos_l0, pdos_l1, pdos_l2 = np.loadtxt("Atom0000001DOS_l.dat",unpack=True,comments=['#','&'])
```

```python
plt.plot(en,pdos,label='Total pDOS')
plt.plot(en,pdos_l0,label='l=0 pDOS')
plt.plot(en,pdos_l1,label='l=1 pDOS')
plt.plot(en,pdos_l2,label='l=2 pDOS')
plt.xlim(-10,10)
plt.ylim(0,14)
plt.xlabel("Energy (eV)")
plt.ylabel("DOS (arbitrary units)")
plt.legend()
#plt.savefig("C1_pDOS.png",dpi=300)
plt.show()
```
