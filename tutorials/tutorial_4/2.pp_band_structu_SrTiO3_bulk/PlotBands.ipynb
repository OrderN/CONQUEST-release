# Plotting band structure for SrTiO$_3$

This Jupyter notebook allows you to plot the band structure produced by CONQUEST and the post-processing.  Reading the file takes most of the work to ensure that it's properly parsed.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### Formatting
The band structure file starts with header lines indicated by the '#' character.  For each band, the energies are then given at each k-point, with bands separated by the '&' character, and a comment line giving the band number.

```python
# Open file and read lines
file = open("BandStructure.dat",mode='r')
lines = file.readlines()

# Count number of bands and k-points
nbands = 0
nkpoints = 0
for line in lines:
    if (line[0]!='#' and line[0]!='&' and nbands==0):
        nkpoints += 1
    elif (line[0]=='&'):
        nbands += 1

# Allocate space for data and read
bands = np.zeros((nbands,nkpoints))
iband = 0
ikpoint = 0
for line in lines:
    # First condition is for data line
    if (line[0]!='#' and line[0]!='&'):
        line = line.split()
        bands[iband,ikpoint] = float(line[1])
        ikpoint += 1
    elif (line[0]=='&'):  # End of band
        ikpoint = 0
        iband += 1
```

```python
# Set limits on bands in eV
emin = -10.0
emax = 10.0

# Create array of labels of special points
ticks = np.array([0,40,80,120])
labels = [r'$\Gamma$','X','M',r'$\Gamma$']

# Plot bands between limits
for i in range(nbands):
    if (np.max(bands[i])<emax and np.min(bands[i])>emin):
        plt.plot(bands[i],'k')
plt.ylabel("Energy (eV)")
plt.xticks(ticks,labels)
plt.grid()
#plt.savefig("BandStructure.png",dpi=300)
plt.show()
```
