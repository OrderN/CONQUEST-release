# Plotting DOS for C$_2$H$_4$

This simple Jupyter notebook allows you to plot the DOS produced by CONQUEST and the post-processing.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Specify & as a comment as the file ends with it
# column 2 shows integrated DOS if you want it
en, dos = np.loadtxt("DOS.dat",unpack=True,comments=['#','&'],usecols=(0,1))
```

```python
plt.plot(en,dos)
plt.xlim(-12.5,12.5)
plt.xlabel("Energy (eV)")
plt.ylabel("DOS (arbitrary units)")
plt.show()
```
