### 3. Operating system <hr style="border:1px solid grey">

- OS makes the link between the *Hardware* and *Software*
- The OS manages the computer ressources wrt. the tasks, that is,
given an amount of work how to share ressources in order to do the job that 
as fast as possible... *optimization of the efficiency*

[//]: <img src="./reference/figures/6757294_700b.jpg" alt="Alternative text" width="500"/>

The OS is in charge of managing
- The memory access and distribution: *Ramdom Access Memory* (RAM)
- The processing unit:  *Central Processing Unit* (CPU)
- Peripheral devices: keyboard, mouse, screen, file system...
- Network: ethernet, wifi, bluetooth
- Administration: user rights, login, pwd...

<hr style="border:1px solid grey">

### 4. Definitions: $\texttt{kernel}/\texttt{shell}/\texttt{bash}/\texttt{terminal}$ <hr style="border:1px solid grey">

[//]: <img src="./reference/figures/shell_layers.png" alt="Alternative text" width="500"/>

- $\texttt{kernel}$: core of the OS, that is, the very first layer controlling the hardware
- $\texttt{shell}$: user interface for access to the OS services ; command line is used for requests
send to the kernel
- $\texttt{bash}$: a command line interpreter (sh, ksh, csh...)
- $\texttt{terminal/console}$: a physical device consisting of a keyboard and a screen, or here, 
a program which emulates a physical terminal
